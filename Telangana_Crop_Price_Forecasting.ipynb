```python
# Cell 2: Upload Dataset
from google.colab import files
uploaded = files.upload()  # Upload daily_prices.csv
```

```python
# Load Data
df = pd.read_csv(list(uploaded.keys())[0])
print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumns:", df.columns.tolist())
```

```python
# Cell 3: Quick EDA
print("Dataset Info:")
print(df.info())
print("\nDate Range:", df['DDate'].min(), "to", df['DDate'].max())
print("\nTop 10 Crops:")
print(df['CommName'].value_counts().head(10))
print("\nTop 5 Yards:")
print(df['YardName'].value_counts().head(5))
print("\nModal Price Stats:")
print(df['Model'].describe())
```

```python
# Cell 4 FIXED: Rice Filter (Copy-Paste & Run)
rice_data = df[df['CommName'].str.contains('Paddy|Dhan|Rice', case=False, na=False)].copy()
print(f"✅ Total Rice Records: {len(rice_data)}")

# Hyderabad yards only
key_yards = rice_data['YardName'].str.contains('HYDERABAD|KARIMNAGAR|WARANGAL', case=False, na=False)
rice_key = rice_data[key_yards].copy()
print(f"✅ Key Yards Rice: {len(rice_key)}")
print("\nSample Rice Data:")
print(rice_key[['DDate', 'YardName', 'Model', 'Arrivals']].head(10))
print("\nRice Yards Available:")
print(rice_key['YardName'].unique())
```

```python
# Cell 5 FIXED
rice_key['DDate'] = pd.to_datetime(rice_key['DDate'], errors='coerce')
rice_hyd = rice_key[rice_key['YardName'].str.contains('HYDERABAD')].sort_values('DDate')

fig = px.line(rice_hyd, x='DDate', y='Model', title='Rice Prices: Hyderabad Yards (Filtered)')
fig.update_layout(xaxis_title="Date", yaxis_title="Modal Price (₹/Quintal)")
fig.show()
```
